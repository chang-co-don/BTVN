Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím
Ví dụ: Ta nhập h = 4:  



```python
h = int(input())
for i in range(1,h+1):
    khoang_trang = " " * (h-i)
    dau_sao = "*" * (2*i-1)
    print(khoang_trang+dau_sao)
```

Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k)
lớn nhất nhưng nhỏ hơn n 

```python
n = int(input())
k = 0
S = 0
while S + k+1 < n:
    k = k+1
    S = S + k
print(f"so k lon nhat la: {k}")
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 

```python
A = []
while True:
    n = int(input())
    A.append(n)
    if n == -1:
        break
if len(A)>=2:
    so_ln = max(A)
    so_nn = min(A)
print(f"so lon nhat trong day la: {so_ln}")
print(f"so nho nhat trong day la: {so_nn}")
```

Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu
chữ số lẻ 


```python
n = int(input())  
chu_so_chan = 0
chu_so_le = 0
while n>0:
    chu_so_cuoi = n%10
    if chu_so_cuoi % 2 == 0:
        chu_so_chan = chu_so_chan +1
    else:
        chu_so_le = chu_so_le +1
    n = n//10
print(f"so chu so chan la: {chu_so_chan}")
print(f"so chu so le la: {chu_so_le}")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số
kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong
dãy số fibonacci lớn nhất nhưng không vượt quá A 


```python
A = int(input())
a = 1
b = 1
while b <=A:
    tong = a + b
    a = b
    b = tong
print(f"{a} la so trong day fibonacci lon nhat thoa man khi nhap vao so {A}")
```

Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố. 

```python
n = int(input())  
n_ban_dau = n
chu_so_nguyen_to = 0
while n>0:
    so_cuoi = n%10
    if so_cuoi in [2, 3, 5, 7]:
        chu_so_nguyen_to = chu_so_nguyen_to + 1
    n = n//10
print(f"{n_ban_dau} co so chu so nguyen to la: {chu_so_nguyen_to}")
  
```
