lệ hay không. Nếu không in ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều", "Tam giác cân", hay
"Tam giác thường". 


```python
a, b ,c = map(int, input().split())
if a+b>c and a+c>b and b+c>a:
    if a == b == c:
        print("Tam Giac Deu")
    elif a==b!=c or a==c!=b or b==c!=a:
        print("Tam giac can")
    else:
        print("Tam giac thuong")
else:
    print("khong phai tam giac")
```

### W3A13
[Tính tiền điện]
Nhập vào số kWh điện tiêu thụ. Tính tiền điện theo các bậc sau:
- 0-50 kWh: 1500đ/kWh
- 51-100 kWh: 2000đ/kWh
- 100 kWh: 3000đ/kWh 


```python
a = float(input())
if 0<= a <= 50:
    tien = a*1500
elif 50 < a <= 100:
    tien = 50*1500 + (a-50)*2000
elif a>100:
    tien = 50*1500 + 50*2000 + (a-100)*3000
print(tien)
```

### W3A16
[Rounding]
Nhập vào một số thực: in ra kết quả làm tròn lên ,tròn xuống, làm tròn tới số nguyên gần nhất (không được sử dụng hàm
sẵn có của python). Các kết quả cách nhau một dấu cách 


```python
n = float(input())
if n==int(n):
    len = int(n)
    xuong = int(n)
    gan_nhat = int(n)
else:
    if n >0:  
       xuong = int(n)
       len = int(n) + 1
       gan_nhat = int(n+0.5)
    else:
       xuong = int(n) - 1
       len = int(n)
       gan_nhat = int(n-0.5)
print(len, xuong, gan_nhat)


```

### W3A17
Cấp số nhân
Cho 4 số a, b, c, d. Hãy kiểm tra xem 4 số này có thể theo thứ tự tạo thành 1 cấp số nhân với công bội nguyên theo đúng
thứ tự a, b, c, d hay không?
Gợi ý : Tìm công bội (b / a) nếu b chia hết cho a, sau đó lấy b nhân công bội và so sánh vs c, c nhân công bội và so sánh vs
d. 

```python
a, b, c, d = map(int, input().split())
k = b/a
if b*k==c and c*k==d:
    print("a b c d tao thanh 1 cong boi nguyen")
else:
    print("No")
```
