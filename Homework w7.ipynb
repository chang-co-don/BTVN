Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
def perfect_arr(lst):
    dict = {}
    count = 0
    for i in lst:
        if i in dict:
            dict[i] += 1
        else:
            dict[i] = 1
    
    for key, value in dict.items():
        if value >= key :
            count += (value - key)
        elif value < key:
            count += value
    if count == 0:
        return "Yes"
    else:
        return f"NO {count}"
    
lst = list(map(int, input().split()))
ket_qua = perfect_arr(lst)
print(ket_qua)
```

W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
def find_majority_element(nums):
    n = len(nums)
    if n == 0:
        return
    candidate = None
    count = 0
    for x in nums:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
    actual_count = 0
    for x in nums:
        if x == candidate:
            actual_count += 1
    if actual_count > n // 2:
        print(candidate)
```

W7A18
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần ; ; 


```python
def bubble_sort(lst):
    n = len(lst)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+1] = lst[j+1], lst[j]
                swapped = True
        if not swapped:
            break
    return lst

lst = list(map(int, input().split()))
day_chan = []
day_le = []
for k in lst:
    if k % 2==0:
        day_chan.append(k)
    if k%2 == 1:
        day_le.append(k)
chan_sort = bubble_sort(day_chan)
le_sort = bubble_sort(day_le)
le_reverse = le_sort[::-1]
result = chan_sort + le_reverse
print(result)



```

Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
lst1 = list(map(int, input().split()))
lst2 = list(map(int, input().split()))
is_ascending = True   # ta mặc định dãy tăng là đúng, loại đi trường hợp dãy rỗng hoặc 1 phần tử
if len(lst1) > 1:
    is_ascending = lst1[0] <= lst1[-1]
elif len(lst2) > 1:
    is_ascending = lst2[0] <= lst2[-1]
# tạo 1 danh sánh rỗng và 2 con trỏ có index 0 ở 2 list
res = []
i = 0
j = 0
while i < len(lst1) and j < len(lst2):
    if is_ascending:
        if lst1[i] <= lst2[j]:
            res.append(lst1[i])
            i += 1
        else:
            res.append(lst2[j])
            j += 1
    else:
        if lst1[i] >= lst2[j]:
            res.append(lst1[i])
            i += 1
        else:
            res.append(lst2[j])
            j += 1
# sau khi chạy đồng thời 2 con trỏ 2 mảng sẽ còn 1 phần tử lớn nhất hoặc bé nhất
#phần tử đó có thể ở lst1 hoặc lst2
# ta chạy cả 2 trường hợp và cài điều kiện dừng
while i < len(lst1):
    res.append(lst1[i])
    i += 1
while j < len(lst2):
    res.append(lst2[j])
    j += 1

print(res)
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
count = {}
for i in range(n):
    full_name = input().strip()
    name = full_name.split()[-1]

    if name in count:
        count[name] += 1
    else:
        count[name] = 1
max_freq = 0
result = ""
for name in count:
    if count[name] > max_freq:
        max_freq = count[name]
        result = name
    elif count[name] == max_freq:
        if name < result:
            result = name
print(result)
```

Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí
tự cái.

```python
s = input().split()
count = 0
for word in s:
    valid = True
    if len(word) < 2:
        valid = False
    elif not ('A' <= word[0] <= 'Z'):
        valid = False

    else:
        for c in word[1:]:
            if not ('a' <= c <= 'z'):
                valid = False
                break
    if valid:
        count += 1
print(count)
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s = input()
stack = []
valid = True
for c in s:
    if c == '(' or c == '[' or c == '{':
        stack.append(c)
    else:
        if len(stack) == 0:
            valid = False
            break

        top = stack[-1]
        stack.pop()

        if c == ')' and top != '(':
            valid = False

        elif c == ']' and top != '[':
            valid = False

        elif c == '}' and top != '{':
            valid = False

if len(stack) != 0:
    valid = False
if valid:
    print("Valid")
else:
    print("Invalid")
```

W7A28
Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True
N = int(input())
best_start = 0
best_len = 0
current_start = 0
current_len = 0

for i in range(1, N + 1):
    if not is_prime(i):
        if current_len == 0:
            current_start = i
        current_len += 1
        if current_len > best_len:
            best_len = current_len
            best_start = current_start
    else:
        current_len = 0
for i in range(best_start, best_start + best_len):
    print(i, end=" ")
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy số phần tử bằng nhau

```python
arr = list(map(int, input().split()))
best = []
current = [arr[0]]
for i in range(1, len(arr)):
    if arr[i] > arr[i - 1]:
        current.append(arr[i])
    else:
        if len(current) > len(best):
            best = current

        current = [arr[i]]
if len(current) > len(best):
    best = current
for x in best:
    print(x, end=" ")
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím


```python
arr = list(map(int, input().split()))
n = len(arr)
dp = [1] * n
trace = [-1] * n
for i in range(n):
    for j in range(i):
        if arr[j] < arr[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            trace[i] = j
max_len = dp[0]
pos = 0
for i in range(1, n):
    if dp[i] > max_len:
        max_len = dp[i]
        pos = i
result = []
while pos != -1:
    result.append(arr[pos])
    pos = trace[pos]

result.reverse()
for x in result:
    print(x, end=" ")
```
