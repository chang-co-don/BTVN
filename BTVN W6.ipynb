đường chéo phụ của mảng

W6A18
Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k) 

```python
data = list(map(int, input().split()))
m, n, k = data[0], data[1], data[2]
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

for row in matrix:
    for item in row:
        print(f"{item:>5}", end="")
    print()

tong_hang = 0
for j in matrix[k-1]:
    tong_hang = tong_hang + j

tong_cot = 0
for row in matrix:
    tong_cot += row[k-1]
print(tong_hang, tong_cot, end=" ")
```

W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def vi_tri(lst,k):
    for i in range(len(lst)):
        if lst[i]==k:
            return i
    return -1

lst = list(map(int, input().split()))
k = int(input())
ket_qua = vi_tri(lst,k)
print(ket_qua)
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện


```python
data = list(input().split())
dic = {}
for i, x in enumerate(data):
    if x in dic:
        dic[x].append(i)
    else:
        dic[x] = [i]
for j in dic:
    dic[j] = tuple(dic[j])
print(data, dic)
```

W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def day_vi_tri(lst):
    lst1 = []
    lst2 = []
    lst_moi = []
    for i in lst:
        if i == 0:
            lst1.append(i)
        else:
            lst2.append(i)
    lst_moi = lst1 + lst2
    return lst_moi

lst = list(map(int, input().split()))
ket_qua = day_vi_tri(lst)
print(lst, ket_qua)
```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
lst = list(map(int, input().split()))
S = 0
for i in range(1,len(lst)-1):
    if lst[i]>lst[i-1] and lst[i]>lst[i+1]:
        S += lst[i]
print(lst, S)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
def check_day(lst):
    day_tang = True
    day_giam = True
    if len(lst)<2: return "day la day vo danh"
    for i in range(len(lst)-1):
        if lst[i+1]-lst[i]>=0:
            day_giam = False
        elif lst[i+1]<=lst[i]:
            day_tang = False
    if day_tang:
        return "day la day tang"
    elif day_giam:
        return "day la day giam"
    else:
        return "day la day vo danh"

lst = list(map(int, input().split()))
ket_qua = check_day(lst)
print(ket_qua)


```
