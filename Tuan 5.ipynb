Viết hàm kiêểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số


```python
def so_chinh_phuong(x):
    if x < 0:
        return False
    
    i = 0
    while i * i <= x:
        if i * i == x:
            return True
        i += 1
    return False
```

VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

```python
def so_doi_xung(x):
    if x < 0:
        return False
    
    original = x
    rev = 0
    
    while x > 0:
        rev = rev * 10 + x % 10
        x //= 10
    
    return original == rev
```

```python
def so_ki_dieu(x):
    if x == 0:
        return False
    
    t = x
    s = 0
    while t > 0:
        s += t % 10
        t //= 10
    
    return x % s == 0
```

W5A20 
Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.


```python
def detect_number_type(s):
    has_dot = False
    i = 0
    while i < len(s):
        if s[i] == '.':
            if has_dot:
                return 'neither'
            has_dot = True
        elif not ('0' <= s[i] <= '9'):
            return 'neither'
        i += 1
    
    if has_dot:
        return 'float'
    return 'int'
```

Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        if b == 0:
            return 0
        return 1
    
    delta = b*b - 4*a*c
    
    if delta < 0:
        return 0
    elif delta == 0:
        return 1
    else:
        return 2
```

W5A25
Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. 
VD: quad = [(0,0), (4,0), (4,3),(0,3)], p = (2,1)


```python
def area(x1,y1,x2,y2,x3,y3):
    a = x1*(y2-y3) + x2*(y3-y1) + x3*(y1-y2)
    if a < 0:
        a = -a
    return a

def point_in_triangle(x1,y1,x2,y2,x3,y3,px,py):
    A = area(x1,y1,x2,y2,x3,y3)
    A1 = area(px,py,x2,y2,x3,y3)
    A2 = area(x1,y1,px,py,x3,y3)
    A3 = area(x1,y1,x2,y2,px,py)
    
    return A == A1 + A2 + A3

def point_in_quadrilateral(quad, p):
    x1,y1 = quad[0]
    x2,y2 = quad[1]
    x3,y3 = quad[2]
    x4,y4 = quad[3]
    px,py = p
    
    return (point_in_triangle(x1,y1,x2,y2,x3,y3,px,py) or
            point_in_triangle(x1,y1,x3,y3,x4,y4,px,py))
```

Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def so_chu_so_khac_0(lst):
    count = 0
    
    i = 0
    while i < len(lst):
        x = lst[i]
        
        if x == 0:
            i += 1
            continue
        
        while x > 0:
            if x % 10 != 0:
                count += 1
            x //= 10
        
        i += 1
    
    return count
```

Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
def sqrt(x):
    i = 0
    while i * i <= x:
        i += 1
    return i - 1  

def angle_ABC(x1, y1, x2, y2, x3, y3):
    BAx = x1 - x2
    BAy = y1 - y2
    BCx = x3 - x2
    BCy = y3 - y2

    dot = BAx * BCx + BAy * BCy

    lenBA = sqrt(BAx*BAx + BAy*BAy)
    lenBC = sqrt(BCx*BCx + BCy*BCy)

    if lenBA == 0 or lenBC == 0:
        return 0

    cos_val = dot / (lenBA * lenBC)

    angle = (1 - cos_val) * 90   # gần đúng

    return angle
```

W5A29 Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min]

```python
def max_min(a, b, c):
    max_v = a
    min_v = a
    
    if b > max_v:
        max_v = b
    if c > max_v:
        max_v = c
    
    if b < min_v:
        min_v = b
    if c < min_v:
        min_v = c
    
    return [max_v, min_v]
```

W5A30 Viết hàm tính hệ số nhị thức C(n, k).


```python
def he_so_nhi_thuc(n, k):
    if k < 0 or k > n:
        return 0
    
    if k > n - k:
        k = n - k
    
    result = 1
    i = 1
    while i <= k:
        result = result * (n - k + i) // i
        i += 1
    
    return result
```

```python
def tong_binh_phuong(n):
    if n <= 0:
        return 0
    
    total = 0
    i = 1
    while i <= n:
        total += i * i
        i += 1
    
    return total
```

Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == "standard":
        base = 10000
        per_km = 12000
    else:
        base = 15000
        per_km = 15000
    
    fare = base + km * per_km
    
    fare += wait_minutes * 1000
    
    if start_hour >= 22 or start_hour < 5:
        fare = fare + fare * 0.2
    
    if km > 10:
        fare = fare - fare * 0.1
    
    return fare
```

W5A34
Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau


```python
def has_unique_digits(x):
    seen = [0]*10
    
    while x > 0:
        d = x % 10
        if seen[d] == 1:
            return False
        seen[d] = 1
        x //= 10
    
    return True

def sum_1_to_n(n):
    if n <= 0:
        return 0
    
    total = 0
    i = 1
    while i <= n:
        if has_unique_digits(i):
            total += i
        i += 1
    
    return total
```
