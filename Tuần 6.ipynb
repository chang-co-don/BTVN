W6A4
Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key
lặp lại, thêm vào list của key đó

```python
data = input().split()
tu_dien = {}
for item in data:
    key, value = item.split(':')
    if key in tu_dien:
        tu_dien[key].append(value)
    else:
        tu_dien[key] = [value]
print(tu_dien)
                        
```

W6A9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một
dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ
được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới
theo thứ tự từ điển tăng dần của key.

```python
def nhap_dict():
    data = input().split()   
    d = {}
    for item in data:
        key, value = item.split(':')
        d[key] = int(value)
    return d
dict1 = nhap_dict()
dict2 = nhap_dict()
dict_tong = dict1.copy()

for key, value in dict2.items():
    if key in dict_tong:
        dict_tong[key] += value
    else:
        dict_tong[key] = value

keys = list(dict_tong.keys())
n = len(keys)
# Bước 2: Sắp xếp nổi bọt (Bubble Sort) cho list keys
for i in range(n):
    for j in range(0, n - i - 1):
        # So sánh hai key đứng cạnh nhau
        if keys[j] > keys[j+1]:
            # Nếu key trước lớn hơn key sau thì hoán đổi (swap)
            keys[j], keys[j+1] = keys[j+1], keys[j]

ket_qua = {}
for k in keys:
    ket_qua[k] = dict_tong[k]

print(ket_qua)
```

W6A12
[Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong
list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ
nhất trong số đó.

```python
lst = list(map(int, input().split()))
dic = {}
for i in lst:
    if i in dic:
        dic[i]+=1
    else:
        dic[i]=1

so_can_tim = 0
max_length = 0
for key, value in dic.items():
    if value > max_length:
        max_length = value
        so_can_tim = key
    elif value == max_length:
        if key < so_can_tim:
            so_can_tim = key
print(lst, dic, so_can_tim, max_length, sep="\n")
```

[Đảo ngược key và value của dictionary]
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo
một dictionary mới bằng cách đảo ngược key và value của dictionary ban

```python
data = input().split()
dict_cu = {}
for i in data:
    key, value = i.split(':')
    dict_cu[key] = value
dict_moi = {}
for key, value in dict_cu.items():
    dict_moi[value] = key
print(dict_cu, dict_moi, sep="\n")
```

W6A15
[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp
key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary
mới.

```python
data = input().split()
k = int(input())
diction = {}
new_dic = {}
for i in data:
    key, value = i.split(':')
    diction[key] = int(value)
for key, value in diction.items():
    if diction[key] > k:
        new_dic[key] = value
print(diction, k, new_dic, sep="\n")

```
